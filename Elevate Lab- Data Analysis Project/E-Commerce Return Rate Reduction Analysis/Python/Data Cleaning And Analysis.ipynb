```python
category_total = df.groupby("Product_Category").size()


category_returned = (
    df[df["Return_Status"] == "Returned"]
    .groupby("Product_Category")
    .size()
)


category_return_rate = (
    category_returned / category_total * 100
).fillna(0)
```

```python
category_analysis = pd.DataFrame({
    "Total_Orders": category_total,
    "Returned_Orders": category_returned,
    "Return_Rate (%)": category_return_rate
}).fillna(0)


print(category_analysis)
```
